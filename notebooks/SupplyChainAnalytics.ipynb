**Data Cleaning**

```python
#Check Duplicate rows
df.duplicated().sum()
```

```python
# Get the null counts
null_counts = df.isnull().sum()

null_counts[null_counts > 0]
```

here the product description column is entirely missing

```python
df.drop(columns=['Product Description'], inplace=True)
```

```python
df.head()
```

```python
#Zipcode is not required for analysis
df.drop(columns=['Order Zipcode'], inplace=True)
```

```python
#remove irrelevent / sensitive colums
df.drop(columns=['Customer Email', 'Customer Fname', 'Customer Id', 'Customer Lname', 'Customer Password', 'Product Image'], inplace=True)
```

```python
df.head()
```

```python
#Convert Date Columns
df['Order Date'] = pd.to_datetime(df['order date (DateOrders)'])

df['Shipping Date'] = pd.to_datetime(df['shipping date (DateOrders)'])
```

**Feature Engineering**

```python
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Order Month Name'] = df['Order Date'].dt.month_name()
df['Order Quarter'] = df['Order Date'].dt.quarter
df['Order Day'] = df['Order Date'].dt.day
```

```python
#Delivery Delay
df['Delivery Delay'] = (df['Days for shipping (real)'] - df['Days for shipment (scheduled)'])

df['Delivery Performance'] = np.select(
    [
        df['Delivery Delay'] < 0,
        df['Delivery Delay'] == 0,
        df['Delivery Delay'] >0
    ],
    [
        'Early',
        'On Time',
        'Delayed'
        ],
    default='Unknown'
)
```

```python
#Profit Margin

df['Profit Margin %'] = (
    df['Order Profit Per Order'] /
    df['Sales']
) * 100
```

**EDA**

```python
#Overall Business Overview

total_orders = df['Order Id'].nunique()
total_sales = df['Sales'].sum()
total_profit = df['Order Profit Per Order'].sum()

print(f'Total Orders: {total_orders}')
print(f'Total Sales: {total_sales:.2f}')
print(f'Total Profit:{total_profit:.2f}')
```

```python
# Sales trend by Year
sales_by_year = df.groupby('Order Year')['Sales'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='Sales', y='Order Year', data=sales_by_year, orient='h')
plt.title('Sales Trend by Year')
plt.xlabel('Total Sales')
plt.ylabel('Years')
plt.grid(axis='x')
plt.show()
```

```python
# Sales trend by Month
sales_by_month = df.groupby('Order Month Name')['Sales'].sum().reset_index()

# Define the order of months for correct plotting
month_order = [
    'January', 'February', 'March', 'April', 'May', 'June',
    'July', 'August', 'September', 'October', 'November', 'December'
]
sales_by_month['Order Month Name'] = pd.Categorical(
    sales_by_month['Order Month Name'], categories=month_order, ordered=True
)
sales_by_month = sales_by_month.sort_values('Order Month Name')


plt.figure(figsize=(12, 7))
sns.barplot(x='Order Month Name', y='Sales', data=sales_by_month)
plt.title('Sales Trend by Month')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

```python
# Sales trend by Quarter
sales_by_quarter = df.groupby('Order Quarter')['Sales'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='Order Quarter', y='Sales', data=sales_by_quarter)
plt.title('Sales Trend by Quarter')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')
plt.grid(axis='y')
plt.show()
```

```python
# Market Analysis
sales_by_region = df.groupby('Order Region')['Sales'].sum().reset_index()

plt.figure(figsize=(12, 7))
sns.barplot(x='Order Region', y='Sales', data=sales_by_region)
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

## Sales vs. Profit by Region

```python
profit_by_region = df.groupby('Order Region')['Order Profit Per Order'].sum().reset_index()

merged_df = pd.merge(sales_by_region, profit_by_region, on='Order Region')

merged_df = merged_df.rename(columns={'Order Profit Per Order': 'Profit'})

plt.figure(figsize=(15, 8))

bar_width = 0.35
index = np.arange(len(merged_df['Order Region']))

# Plotting Sales bars
plt.bar(index - bar_width/2, merged_df['Sales'], bar_width, label='Total Sales', color='skyblue')

# Plotting Profit bars
plt.bar(index + bar_width/2, merged_df['Profit'], bar_width, label='Total Profit', color='lightcoral')

plt.title('Total Sales and Profit by Region')
plt.xlabel('Order Region')
plt.ylabel('Amount')
plt.xticks(index, merged_df['Order Region'], rotation=60, ha='right')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(x='Sales', y='Profit', hue='Order Region', data=merged_df, s=100, alpha=0.8)

plt.title('Sales vs. Profit by Order Region')
plt.xlabel('Total Sales')
plt.ylabel('Total Profit')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## Delivery Performance Analysis by Region

```python
delivery_performance_by_region = df.groupby(['Order Region', 'Delivery Performance']).size().unstack(fill_value=0)

# Sort the columns for consistent plotting (e.g., Early, On Time, Delayed)
delivery_performance_by_region = delivery_performance_by_region[['Early', 'On Time', 'Delayed']]

delivery_performance_by_region.plot(kind='bar', stacked=True, figsize=(15, 8))
plt.title('Delivery Performance by Order Region')
plt.xlabel('Order Region')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Delivery Status')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
sales_by_department = df.groupby('Department Name')['Sales'].sum().nlargest(5).reset_index()

plt.figure(figsize=(10, 8))
plt.pie(sales_by_department['Sales'],
        labels=sales_by_department['Department Name'],
        autopct='%1.1f%%',
        startangle=140,
        explode=[0.05]*len(sales_by_department),
        shadow=True)
plt.title('Top 5 Departments by Sales')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
profit_margin_by_department = df.groupby('Department Name')['Profit Margin %'].mean().sort_values(ascending=False).reset_index()

plt.figure(figsize=(12, 7))
sns.barplot(x='Profit Margin %', y='Department Name', data=profit_margin_by_department, palette='viridis')
plt.title('Average Profit Margin % by Department Name')
plt.xlabel('Average Profit Margin (%)')
plt.ylabel('Department Name')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
#Delivery Analysis
delivery_rate = (
    df.groupby('Shipping Mode')['Late_delivery_risk']
    .mean()
    .mul(100)
)
```

```python
delivery_rate.plot(kind='barh', figsize=(10, 6))
plt.title('Delivery Risk by Shipping Mode')
plt.ylabel('Shipping Mode')
plt.xlabel('Delivery Risk (%)')
```

## Correlation Heatmap

```python
selected_columns = [
    'Days for shipping (real)',
    'Days for shipment (scheduled)',
    'Benefit per order',
    'Sales per customer',
    'Late_delivery_risk',
    'Order Item Discount',
    'Order Item Product Price',
    'Sales',
    'Order Item Total',
    'Product Price',
    'Profit Margin %'
]

# Filter the DataFrame to include only the selected columns
df_selected_numerical = df[selected_columns]

# Calculate the correlation matrix
correlation_matrix = df_selected_numerical.corr()

plt.figure(figsize=(14, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Selected Numerical Variables')
plt.show()
```
